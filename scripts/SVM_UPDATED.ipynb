```python
# Loading the necessary libraries: 

%pip install imbalanced-learn

import pandas as pd
import matplotlib.pyplot as plt                                                                       # For the SVM, AUROC, and AUPRC graphs. 
import seaborn as sns                                                                                 # For the correlation/feature selection heatmap. 
from sklearn.model_selection import train_test_split                                                  # For the SVM models. 
from sklearn.preprocessing import StandardScaler                                                      # For the SVM models. 
from sklearn.metrics import roc_auc_score                                                             # For the AUROC graph. 
from sklearn.svm import SVC                                                                           # For the SVM models. 
from sklearn import metrics                                                                           # For evaluating the metrics from SVM and AUROC/AUPRC. 
from sklearn.inspection import DecisionBoundaryDisplay                                                # For the SVM graphs. 
from sklearn.metrics import roc_curve, precision_recall_curve, average_precision_score                # For the AUROC and AUPRC graphs respectively. 
from imblearn.over_sampling import SMOTE                                                              # For balancing the dataset before generating SVM graphs. 
```

```python
# Loading the clinical dataset and data exploration: 

diabetes_data = pd.read_csv("C:/Users/Zahin/Desktop/Semester 2/Machine Learning and AI/diabetes_health_indicators.csv")  
diabetes_data.head()
diabetes_data.info()
diabetes_data.describe()
print(diabetes_data.isnull().sum())
print(diabetes_data.columns)
print(diabetes_data.shape)
```

```python
# Finding the features in the dataset: 

data_columns = diabetes_data.columns.to_list()                                  # Added the columns from the diatetes dataset to a list. 
print("Features: ", data_columns)

# Plotting a correlation/feature selection heatmap:

data_corr_matrix = diabetes_data.corr()
print(data_corr_matrix)
plt.figure(figsize=(18, 16))
sns.heatmap(data_corr_matrix, annot=True, cmap="coolwarm", cbar=True, square=True, linewidths=2, linecolor="Black")
plt.title("Feature Selection Heatmap")
plt.tight_layout()
plt.show()

# Feature selection: determining the correlations with the target column, Diabetes_binary: 

target_col = data_corr_matrix["Diabetes_binary"].drop("Diabetes_binary")
feature_selection = target_col.abs().sort_values(ascending=False)
top_10_features = feature_selection.head(10).index.to_list()                  # The top features from the diabetes dataset were chosen for further analysis. 
print(top_10_features)
```

```python
# Support Vector Machines (SVM) Model (retrieved codes from: https://www.datacamp.com/tutorial/svm-classification-scikit-learn-python): 

# Splitting the dataset: 

from sklearn.svm import LinearSVC

X = diabetes_data[top_10_features]
y = diabetes_data["Diabetes_binary"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3,random_state=42)  # To test 30% and train 70% of the dataset (as mentioned in the aforementioned link). 

# Standardization of the data:

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Balancing the dataset. The SMOTE feature was chosen and codes were retrieved from (https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html):

balanced = SMOTE(random_state=42)                                                    
X_train_balanced, y_train_balanced = balanced.fit_resample(X_train_scaled, y_train)

# Creating an SVM Classifier (linear kernel). The codes for the linear SVC were retrieved from (https://scikit-learn.org/stable/modules/generated/sklearn.svm.LinearSVC.html):

svm_clf = LinearSVC(max_iter=10000, random_state=42)     # The max iteration was set to 10,000. This is because the diabetes dataset had points in the 100,000s range.                             
                                                         # Initially I tried without setting the max iteration; however, the model took very long to generate the accuracy,                    
# Training the model using the training sets:            # precision, and recall scores (> 10 minutes). Hence, I added the max_iter argument. 

svm_clf.fit(X_train_balanced, y_train_balanced)

# Predicting the response for the test dataset:
 
y_pred = svm_clf.predict(X_test_scaled)

# Model Accuracy: 

print("Accuracy:",metrics.accuracy_score(y_test, y_pred))

# Model Precision: 

print("Precision:",metrics.precision_score(y_test, y_pred))

# Model Recall: 

print("Recall:",metrics.recall_score(y_test, y_pred))
```

```python
# Plotting the Linear SVM Model (link: https://scikit-learn.org/stable/auto_examples/svm/plot_svm_kernels.html): 

# Chose two of the highest correlated features from the dataset: 

from sklearn import svm

X1 = diabetes_data[["BMI", "Age"]].values          # The BMI and Age values were chosen in this case. Other selected features were substituted. However, they 
                                                   # did not show any significant result changes (i.e. points overlapped and lines were all horizontal). 
y1 = diabetes_data["Diabetes_binary"].values

X1, _, y1, _ = train_test_split(X1, y1, train_size=10000, random_state=42)  # Again, the train_size was set to 10,000 because the model was taking > 10 minutes to load. 

# Scaling the features:

scaler1 = StandardScaler()
X1_scaled = scaler1.fit_transform(X1)

# Balancing the dataset:

balanced = SMOTE(random_state=42)
X1_train_balanced, y1_train_balanced = balanced.fit_resample(X1_scaled, y1)

# Defining the plot training data and training the SVM Model: 

def plot_training_data_with_decision_boundary(kernel="linear", ax=None, long_title=True, support_vectors=True):
    clf_1 = SVC(kernel=kernel, gamma=0.01)                                # Gamma values of 0.01, 0.1, 1.0, and 2 were tested. However, the overlapping of the points remained. 
    clf_1.fit(X1_train_balanced, y1_train_balanced)

    # Plotting settings:
    
    fig, ax = plt.subplots(figsize=(6, 5))
    
    # Plot decision boundary
    DecisionBoundaryDisplay.from_estimator(
        clf_1, X1_train_balanced, ax=ax,
        response_method="predict",
        plot_method="pcolormesh",
        alpha=0.3
    )
    DecisionBoundaryDisplay.from_estimator(
        clf_1, X1_train_balanced, ax=ax,
        response_method="decision_function",
        plot_method="contour",
        levels=[-1, 0, 1],
        colors=["k", "k", "k"],
        linestyles=["--", "-", "--"]
    )
    
    ax.scatter(clf_1.support_vectors_[:, 0], clf_1.support_vectors_[:, 1],
                   s=150, facecolors='none', edgecolors='k')

    # Plotting the SVM Model: 
    scatter = ax.scatter(X1_train_balanced[:, 0], X1_train_balanced[:, 1], c=y1_train_balanced, s=30, edgecolors="k")
    ax.legend(*scatter.legend_elements(), loc="upper right", title="Classes")
    ax.set_title(f" Decision boundaries of {kernel} kernel in SVM")
    plt.tight_layout()
    plt.show()
 
plot_training_data_with_decision_boundary("linear")  
```

```python
# Plotting the Radial Basis Function SVM Model (link: https://scikit-learn.org/stable/auto_examples/svm/plot_svm_kernels.html): 
# Since the aforementioned SVM model (with the linear kernel), did not provide satisfactory results, I thought that the SVM model (with the RBF kernel) will be able to capture
# any non-linear relationships between the selected variables from the diabetes dataset. Hence, I reran the SVM model (with the RBF kernel). 
# Chose two of the highest correlated features from the dataset: 

from sklearn import svm

X1 = diabetes_data[["BMI", "Age"]].values                  # The BMI and Age values were chosen in this case. Other selected features were substituted too.
y1 = diabetes_data["Diabetes_binary"].values

X1, _, y1, _ = train_test_split(X1, y1, train_size=10000, random_state=42) # Again, the train_size was set to 10,000 because the model was taking > 10 minutes to load. 

# Scaling the features:

scaler1 = StandardScaler()
X1_scaled = scaler1.fit_transform(X1)

# Balancing the dataset. The SMOTE feature was chosen and codes were retrieved from (https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html):

balanced = SMOTE(random_state=42)
X1_train_balanced, y1_train_balanced = balanced.fit_resample(X1_scaled, y1)

# Defining the plot training data and training the SVM Model: 

def plot_training_data_with_decision_boundary(kernel="rbf", ax=None, long_title=True, support_vectors=True):
    clf_1 = SVC(kernel=kernel, gamma=0.05)                             # Gamma values of 0.05, 0.1, 1.0, and 10 were tested. However, the overlapping of the points remained.       
    clf_1.fit(X1_train_balanced, y1_train_balanced)

    # Plotting settings:
    
    fig, ax = plt.subplots(figsize=(6, 5))
    
    # Plot decision boundary
    DecisionBoundaryDisplay.from_estimator(
        clf_1, X1_train_balanced, ax=ax,
        response_method="predict",
        plot_method="pcolormesh",
        alpha=0.3
    )
    DecisionBoundaryDisplay.from_estimator(
        clf_1, X1_train_balanced, ax=ax,
        response_method="decision_function",
        plot_method="contour",
        levels=[-1, 0, 1],
        colors=["k", "k", "k"],
        linestyles=["--", "-", "--"]
    )
    
    ax.scatter(clf_1.support_vectors_[:, 0], clf_1.support_vectors_[:, 1],
                   s=150, facecolors='none', edgecolors='k')

    # Plotting the SVM Model: 
    scatter = ax.scatter(X1_train_balanced[:, 0], X1_train_balanced[:, 1], c=y1_train_balanced, s=30, edgecolors="k")
    ax.legend(*scatter.legend_elements(), loc="upper right", title="Classes")
    ax.set_title(f" Decision boundaries of {kernel} kernel in SVM")
    plt.tight_layout()
    plt.show()
 
plot_training_data_with_decision_boundary("rbf")  
```
